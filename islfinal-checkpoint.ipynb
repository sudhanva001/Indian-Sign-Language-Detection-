```python
# =========================
# ISL to Text & Speech - Fixed Complete Notebook
# =========================

# --------- STEP 0: Notes before running ----------
# 1) Make sure your dataset is EXTRACTED and has structure:
#    dataset_path/
#       A/
#         img1.jpg
#         ...
#       B/
#       ...
#    (26 folders A..Z assuming letters)
# 2) Set dataset_path below to the extracted folder location.
# 3) Install missing packages if needed:
#    !pip install tensorflow matplotlib opencv-python pyttsx3
# 4) This notebook uses transfer learning (MobileNetV2) by default for better generalization.
#    If your machine cannot download pretrained weights, switch to the simple CNN (comment provided).

# --------- Imports ----------
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2
import pyttsx3
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from collections import deque
```

```python
# UPDATE this to your dataset folder (extracted .rar content)
dataset_path = r"C:\Users\Dell\Downloads\original_images"  # <- change this

# Image size used for training/prediction (MobileNetV2 prefers >= 96; we use 128)
IMG_SIZE = (128, 128)
BATCH_SIZE = 32
EPOCHS = 20
MODEL_SAVE_PATH = "isl_mobilenetv2.h5"
USE_TRANSFER_LEARNING = True   # True -> MobileNetV2; False -> custom small CNN
```

```python
assert os.path.exists(dataset_path), f"Dataset folder not found: {dataset_path}"
classes = sorted([d for d in os.listdir(dataset_path) if os.path.isdir(os.path.join(dataset_path, d))])
print("Found classes (folders):", classes)
num_classes = len(classes)
print("Number of classes:", num_classes)
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.12,
    height_shift_range=0.12,
    shear_range=0.12,
    zoom_range=0.12,
    horizontal_flip=False,   # flipping may not be correct for sign language; set False if signs are hand-specific
    validation_split=0.2
)

train_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='sparse',
    subset='training',
    shuffle=True
)

val_gen = train_datagen.flow_from_directory(
    dataset_path,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='sparse',
    subset='validation',
    shuffle=False
)

# Keep a mapping from index -> label for decoding predictions later
index_to_label = {v: k for k, v in train_gen.class_indices.items()}
print("class_indices:", train_gen.class_indices)
```

```python
if USE_TRANSFER_LEARNING:
    base = MobileNetV2(weights='imagenet', include_top=False, input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
    base.trainable = False  # freeze base for initial training
    x = base.output
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)
    model = models.Model(inputs=base.input, outputs=outputs)
else:
    # Simple CNN fallback (if no internet to download weights)
    model = models.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3)),
        layers.MaxPooling2D(2,2),
        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D(2,2),
        layers.Conv2D(128, (3,3), activation='relu'),
        layers.MaxPooling2D(2,2),
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
# Setup TTS
engine = pyttsx3.init()
engine.setProperty('rate', 150)
voices = engine.getProperty('voices')
# Use voices[0] or voices[1] depending on availability
engine.setProperty('voice', voices[0].id)

def speak(text):
    engine.say(text)
    engine.runAndWait()

# Smoothing queue
QUEUE_LEN = 8
pred_queue = deque(maxlen=QUEUE_LEN)

cap = cv2.VideoCapture(0)    # change index if multiple cameras
if not cap.isOpened():
    raise RuntimeError("Could not open webcam. Check camera index.")

print("Press 'q' in the camera window to quit.")
recent_spoken = None
try:
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frame = cv2.flip(frame, 1)  # mirror for natural feel

        # Define ROI box — adjust coordinates as needed
        h, w = frame.shape[:2]
        box_size = min(h, w) // 2
        cx, cy = w // 2, h // 2
        x1, y1 = cx - box_size//2, cy - box_size//2
        x2, y2 = cx + box_size//2, cy + box_size//2

        roi = frame[y1:y2, x1:x2]
        if roi.size == 0:
            continue

        # Preprocess (BGR->RGB, resize, normalize)
        roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)
        inp = cv2.resize(roi_rgb, IMG_SIZE)
        inp = inp.astype('float32') / 255.0
        inp = np.expand_dims(inp, axis=0)

        preds = model.predict(inp)  # shape (1, num_classes)
        pred_prob = preds[0]
        pred_idx = np.argmax(pred_prob)
        pred_label = index_to_label[pred_idx]
        pred_conf = pred_prob[pred_idx]

        # Add to smoothing queue
        pred_queue.append(pred_idx)
        # majority vote in queue
        most_common = max(set(pred_queue), key=pred_queue.count)
        displayed_label = index_to_label[most_common]

        # Draw UI
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0,255,0), 2)
        cv2.putText(frame, f"ISL Sign: {displayed_label} ({pred_conf:.2f})",
                    (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0,255,0), 2)

        cv2.imshow("ISL Recognition", frame)

        # Speak only when stable and changed
        if displayed_label != recent_spoken and pred_conf > 0.6:
            speak(f"The predicted sign is {displayed_label}")
            recent_spoken = displayed_label

        # Quit on 'q'
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```
